The `pre_processing` function is designed to clean and format the input text for enhanced performance in natural language processing tasks. It performs several key operations to ensure the text is uniformly structured. Firstly, it adds spaces around specific punctuation marks, such as `۔`, `؟`, and `!`, which helps clearly delineate words from punctuation for accurate sentence segmentation. The function also removes curly quotes (i.e., `‘` and `’`), standardizing the appearance of quotation marks to avoid complications in tokenization. Additionally, it consolidates multiple spaces into a single space and trims any leading or trailing whitespace to produce a cleaner text output. Finally, spaces are added around digits to separate them from surrounding text, ensuring that numbers are treated as distinct tokens in subsequent processing steps. Overall, this function addresses common formatting issues, facilitating more accurate and efficient NLP analyses.

```python
def pre_processing(text):
    # Add space before and after specific punctuation
    text = re.sub(r'([۔؟!])([^ ١٢٣٤٥٦٧٨٩٠])', r'\1 \2', text)  # Space after punctuation
    text = re.sub(r'([^ ١٢٣٤٥٦٧٨٩٠])([۔؟!])', r'\1 \2', text)  # Space before punctuation
    # Remove curly quotes
    text = re.sub(r'[‘’]', ' ', text)
    # Remove extra spaces and ensure single space between words
    text = re.sub(r'\s+', ' ', text).strip()
    # Add space around digits
    text = re.sub(r'(\d)', r' \1 ', text)
    return text
```

Technique 1

This code snippet implements an enhanced custom sentence tokenizer for Urdu text, aiming to improve sentence boundary detection. It uses the `re` library for regular expressions and defines sets of pre-defined Urdu conjunctions and sentence-ending words to guide its logic.

In the `improved_custom_sentence_tokenizer` function, the text is first cleaned to normalize spaces around punctuation marks. The tokenizer processes the text word by word, utilizing contextual cues from the predefined conjunctions and sentence-ending words to determine when to finalize a sentence. This ensures that sentences are not prematurely broken, enhancing the overall segmentation accuracy.

The code also provides an example of how to use the tokenizer with a sample Urdu text, allowing for a comparison of the results to evaluate improvements in sentence segmentation.

```python
import re
import urduhack


# Pre-defined Urdu conjunctions and sentence endings
urdu_conjunctions = {'جنہیں', 'جس', 'جن', 'جو', 'اور', 'اگر', 'اگرچہ', 'لیکن', 'مگر', 'پر', 'یا', 'تاہم', 'کہ', 'کر',
                     'تو', 'گے', 'گی', 'جسے', 'تاکہ', 'اسے', 'جب', 'وہ', 'جہاں', 'جبکہ', 'بلکہ', 'کیونکہ'}
urdu_endings = {'کیجیے', 'کیجئے', 'گئیں', 'تھیں', 'ہوں', 'خریدا', 'گے', 'ہونگے', 'گا', 'چاہیے', 'ہوئیں', 'گی',
                'تھا', 'تھی', 'تھے', 'ہیں', 'ہے', 'جاسکے', 'بنیں', 'گئی', 'گیا', 'دیا', 'جائیں', 'چاہئے', 'کھایا', 'رہا', 'سکے', 'کرسکے'}

def improved_custom_sentence_tokenizer(text):
    '''
    Tokenizes the text into sentences using advanced context-aware logic.
    '''
    # Normalize spaces around punctuation
    text = re.sub(r'([۔؟!])', r' \1 ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenize using urduhack as a baseline
    urduhack.download()
    from urduhack.tokenization import sentence_tokenizer
    sentences = sentence_tokenizer(text)

    processed_sentences = []
    current_sentence = []

    for word in text.split():
        word = word.strip()

        # Add new handling for conjunctions and endings
        if word in {'۔', '؟', '!'}:
            if current_sentence:
                processed_sentences.append(' '.join(current_sentence).strip())
                current_sentence = []
            continue

        current_sentence.append(word)

        # Check for contextual clues before breaking a sentence
        if (word in urdu_endings and
            (len(current_sentence) > 1 and current_sentence[-2] not in urdu_conjunctions) and
            (current_sentence[-1] not in urdu_conjunctions)):
            processed_sentences.append(' '.join(current_sentence).strip())
            current_sentence = []

    if current_sentence:
        processed_sentences.append(' '.join(current_sentence).strip())

    return sentences, processed_sentences

# Tokenizing using the enhanced custom tokenizer
urduhack_sentences, enhanced_custom_sentences = improved_custom_sentence_tokenizer(text)

# Output for comparison
print("Urduhack Sentences:", urduhack_sentences)
print("Enhanced Custom Tokenizer Sentences:", enhanced_custom_sentences)
```

This code snippet utilizes the SentencePiece model for tokenizing Urdu text into sentences. The SentencePiece library is designed for subword tokenization, making it effective for handling various linguistic challenges.

In the `sentencepiece_tokenizer` function, a pre-trained SentencePiece model is loaded, allowing the tokenizer to convert the input text into a sequence of tokens. After tokenization, the tokens are joined to simulate sentence reconstruction. The function then uses regular expressions to split the tokenized text at punctuation marks, creating a list of sentences.

Finally, the punctuation is combined with the corresponding sentences, ensuring proper formatting. The output provides a list of sentences segmented by the tokenizer, which can be useful for further text analysis or processing tasks.

```python
import sentencepiece as spm
import re

# Load trained model
sp = spm.SentencePieceProcessor()
sp.load('urdu_model.model')  # Loading model file

def sentencepiece_tokenizer(text):
    '''
    Tokenizes the text using a SentencePiece model.
    '''
    # Tokenize the input text using SentencePiece
    tokenized_sentences = sp.encode(text, out_type=str)

    # Join the tokens to recreate sentences (this simulates a sentence tokenizer)
    processed_sentences = ' '.join(tokenized_sentences)

    # Assuming we split sentences based on special tokens or punctuation
    sentence_list = re.split(r'([۔؟!])', processed_sentences)

    # Combine punctuation with sentences
    final_sentences = []
    for i in range(0, len(sentence_list) - 1, 2):
        final_sentences.append(sentence_list[i].strip() + sentence_list[i + 1].strip())

    return final_sentences

sentencepiece_sentences = sentencepiece_tokenizer(text)
print("SentencePiece Tokenizer Sentences:", sentencepiece_sentences)
```

#### SentencePiece encode() and decode() functions:

This code snippet implements a custom tokenization and encoding system for text processing. It establishes a mapping between tokens and unique identifiers, facilitating the conversion of text into a format suitable for various natural language processing tasks.

The `get_token_id` function generates a unique ID for each new token, storing the mappings in two dictionaries: `token_to_id` for mapping tokens to IDs, and `id_to_token` for the reverse. The `encode` function tokenizes the input text by splitting it based on whitespace and punctuation, then converts these tokens into their corresponding unique IDs using the previously defined function.

The `decode` function reconstructs the original text from the encoded IDs by looking up the tokens in the `id_to_token` dictionary. It handles spacing appropriately, ensuring the reconstructed text retains its original structure.

This approach allows for efficient processing of text, making it easier to manage tokenization and maintain relationships between tokens and their identifiers. The output provides both the encoded IDs and the reconstructed text, demonstrating the functionality of the encoding and decoding process.

```python
import re

# Dictionary to simulate a token-to-id and id-to-token mapping
token_to_id = {}
id_to_token = {}

# This function will generate a unique ID for each new token
def get_token_id(token):
    if token not in token_to_id:
        new_id = len(token_to_id) + 1
        token_to_id[token] = new_id
        id_to_token[new_id] = token
    return token_to_id[token]

def encode(text):
    '''
    Custom encoding function for tokenizing the text.
    It assigns unique IDs to tokens and subwords.
    '''
    # Simple tokenization: split by whitespace and punctuation
    # Adjust the regex pattern to handle punctuation and spaces
    tokens = re.findall(r'\w+|[۔؟!,.]', text)

    # Convert tokens to corresponding IDs
    encoded_ids = [get_token_id(token) for token in tokens]

    return encoded_ids

def decode(encoded_ids):
    '''
    Custom decoding function that reconstructs the text from token IDs.
    '''
    # Convert IDs back to tokens
    decoded_tokens = [id_to_token[token_id] for token_id in encoded_ids]

    # Rebuild the original sentence
    # Handle cases for spacing before punctuation and after words
    text = ''.join([' ' + token if token.isalnum() else token for token in decoded_tokens])

    return text.strip()  # Strip any extra leading/trailing spaces

# Encoding the text into IDs
encoded_text = encode(text)
print("Encoded Text:", encoded_text)

# Decoding the IDs back to the original text
decoded_text = decode(encoded_text)
print("Decoded Text:", decoded_text)
```
